# Setup Test

Run each cell in order. If the last cell prints a response, your environment is working.

```python
import os
from pathlib import Path

from dotenv import load_dotenv

env_path = Path.cwd() / ".env"
if not env_path.is_file():
    env_path = Path.cwd().parent / ".env"
if env_path.is_file():
    load_dotenv(dotenv_path=env_path)

if not os.getenv("OPENAI_API_KEY") or os.getenv("OPENAI_API_KEY") == "your-openai-key-here":
    raise ValueError(f"Set OPENAI_API_KEY in {env_path} before continuing.")
```

```python
from openai import OpenAI

openai_client = OpenAI()
```

```python
response = openai_client.responses.create(
    model="gpt-4o-mini",
    input="Say hello in one short sentence."
)

print(response.output_text)
```
